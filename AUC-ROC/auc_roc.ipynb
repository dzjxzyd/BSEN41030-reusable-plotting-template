## Step 1 - Choose the input and plot

The default demo file contains 100 samples and four model probability columns. The plotting function automatically detects every column except `sample_id` and `true_label` as a model. Two more demo files are included: `auc_demo_data_3_models.csv` and `auc_demo_data_7_models.csv`.

Visual settings: DejaVu Sans, 15 pt body text, normal-weight labels, a 1.5 pt four-sided axis frame, a distinct color for each of up to 10 models, and a 600-dpi white-background PNG. When `title` is `None`, the title is generated from the number of models, for example `ROC Curves for 7 Models`.

```python
from IPython.display import Image, display

# Replace this with another wide-format CSV when needed.
filename = 'auc_demo_data.csv'
title = None        # None creates 'ROC Curves for <n> Models'
output_path = None  # None creates <input>_roc.png

status = main(filename, output_path, title)
if status == 0:
    png_path = Path(output_path) if output_path else Path(filename).with_name(f"{Path(filename).stem}_roc.png")
    display(Image(filename=str(png_path), width=700))
```

## Interpreting the result

- The x-axis is the false-positive rate.
- The y-axis is the true-positive rate.
- AUC is the area under the ROC curve; larger values indicate stronger ranking discrimination on this dataset.
- The dashed diagonal is the random-classifier reference.

The cell creates two files beside the input CSV:

- `<input>_roc.png`
- `<input>_roc_summary.csv`
