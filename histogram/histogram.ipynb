# Multi-column histogram

This notebook uses `histogram.py` in the same folder to load a CSV or Excel file, identify numeric columns, and create a publication-style 600-dpi histogram. Selected columns are overlaid with a colorblind-friendly palette and share the same bin edges, so their bar heights can be compared directly.

If `filename` is set to `None`, the configuration cell will ask for the data-file path.

Prerequisites: basic Python, pandas DataFrames, and simple matplotlib plotting.

````python
from pathlib import Path
import sys

from IPython.display import Image, display

# The notebook and histogram.py should be in the same folder.
NOTEBOOK_DIR = Path.cwd()
SCRIPT_PATH = NOTEBOOK_DIR / "histogram.py"
if not SCRIPT_PATH.exists():
    raise FileNotFoundError(
        f"Could not find {SCRIPT_PATH}. Open this notebook from the histogram folder."
    )

if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

from histogram import (
    create_histogram,
    derive_output_path,
    load_data,
    numeric_columns,
)
````

````python
# Use the bundled example, or replace this with another CSV/Excel path.
# Other demos: "food_antioxidant_capacity_synthetic.csv", "health_systolic_blood_pressure_synthetic.csv"
filename = "histogram_prediction_scores.csv"
# filename = None  # None asks for the data-file path

selected_columns = None  # None selects all numeric columns
bins = 40

if filename is None:
    filename = input("Enter data file path: ").strip()
    if not filename:
        raise ValueError("No data file path was provided.")

data_path = Path(filename).expanduser()
frame = load_data(data_path)
print(f"Loaded {len(frame):,} rows and {frame.shape[1]} columns from {data_path}.")
````

````python
display(frame.head())

available_columns = numeric_columns(frame)
if not available_columns:
    raise ValueError("No numeric columns are available for plotting.")

print("Available numeric columns:")
for number, column in enumerate(available_columns, start=1):
    print(f"  {number}. {column}")

if selected_columns is None:
    selected_columns = available_columns
print(f"Plotting columns: {', '.join(selected_columns)}")
````

## Create the histogram

The output PNG is saved beside the input file, matching the behavior of `histogram.py`. The chart uses the script's reference-aligned typography, axes, line weights, and 600-dpi export settings.

````python
output_path = derive_output_path(data_path)
histogram_path = create_histogram(
    frame,
    selected_columns,
    output_path,
    bins=bins,
    show=False,
)

print(f"Histogram saved to: {histogram_path}")
display(Image(filename=str(histogram_path), width=800))
````
